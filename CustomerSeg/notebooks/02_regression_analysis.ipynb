# 📈 Unit II: Regression Analysis

## 1. Executive Summary
In this unit, we focus on **predicting customer value** (Monetary) based on their behavior (Recency and Frequency). This helps in estimating **Customer Lifetime Value (CLV)**.

**Objectives:**
1.  **Linear Regression**: Establish a baseline model.
2.  **Polynomial Regression**: Capture non-linear relationships.
3.  **Evaluation**: Compare models using MAE, RMSE, and R².

---

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sys
import os

# Add src to path
sys.path.append(os.path.abspath(os.path.join('..')))
from src.models.regression import SalesRegressor

# Load processed RFM data
rfm = pd.read_csv('../data/processed/rfm_customer_data.csv')
print("✅ RFM Data Loaded")
print(rfm.head())
```

## 2. Data Preparation
We use **Recency** and **Frequency** as independent variables (X) to predict **Monetary** value (y).

```python
reg_model = SalesRegressor()
X_train, X_test, y_train, y_test = reg_model.prepare_data(rfm)

print(f"Training Samples: {X_train.shape[0]}")
print(f"Testing Samples: {X_test.shape[0]}")
```

## 3. Linear Regression
The simplest approach assuming a straight-line relationship.

```python
model_lin = reg_model.train_linear(X_train, y_train)
y_pred_lin = reg_model.predict(X_test, model_type='linear')

metrics_lin = reg_model.evaluate(y_test, y_pred_lin)
print("Linear Regression Performance:", metrics_lin)
```

## 4. Polynomial Regression (Degree 2)
Since spending behavior often grows exponentially (e.g., highly frequent buyers spend disproportionately more), a polynomial model might fit better.

```python
model_poly = reg_model.train_polynomial(X_train, y_train, degree=2)
y_pred_poly = reg_model.predict(X_test, model_type='polynomial')

metrics_poly = reg_model.evaluate(y_test, y_pred_poly)
print("Polynomial Regression Performance:", metrics_poly)
```

## 5. Model Comparison & Conclusion
Comparing the two models using the actual vs predicted values.

```python
res_df = pd.DataFrame({
    'Model': ['Linear', 'Polynomial'],
    'MAE': [metrics_lin['MAE'], metrics_poly['MAE']],
    'RMSE': [metrics_lin['RMSE'], metrics_poly['RMSE']],
    'R2': [metrics_lin['R2'], metrics_poly['R2']]
})

print(res_df)

# Visualization of predictions vs actuals (subset for clarity)
plt.figure(figsize=(10, 6))
plt.scatter(y_test, y_pred_lin, alpha=0.5, label='Linear Predictions', color='blue')
plt.scatter(y_test, y_pred_poly, alpha=0.5, label='Polynomial Predictions', color='red')
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'k--', lw=2)
plt.xlabel('Actual Monetary Value')
plt.ylabel('Predicted Monetary Value')
plt.title('Actual vs Predicted Sales')
plt.legend()
plt.show()
```

**📝 Insight:**
- **R² Score**: Indicates how much variance is explained. Higher is better.
- If Polynomial R² > Linear R², it confirms the non-linear nature of customer spending.
- **Recommendation**: Use the model with the higher R² and lower RMSE for CLV prediction.
