## 1. Executive Summary
This unit covers advanced techniques:
1.  **PCA (Principal Component Analysis)**: Reducing dimensionality to visualize high-dimensional data.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sys
import os

# Add src to path
sys.path.append(os.path.abspath(os.path.join('..')))
from src.models.advanced import AdvancedModels
from src.models.regression import SalesRegressor

# Load data
rfm = pd.read_csv('../data/processed/rfm_customer_data.csv')
reg_model = SalesRegressor()
X_train, X_test, y_train, y_test = reg_model.prepare_data(rfm)
```

## 2. Principal Component Analysis (PCA)
Reducing our features to 2 components for visualization.

```python
adv_model = AdvancedModels()
X_pca, variance = adv_model.apply_pca(X_train, n_components=2)

print(f"Explained Variance Ratio: {variance}")
print(f"Total Variance Explained: {sum(variance):.2f}")

plt.figure(figsize=(8, 6))
plt.scatter(X_pca[:, 0], X_pca[:, 1], alpha=0.5, c='purple')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('PCA Visualization of Customer Data')
plt.show()
```

## 3. Neural Network (MLP) Regression
Predicting Sales using a Multi-Layer Perceptron.

```python
mlp_reg = adv_model.train_mlp_regression(X_train, y_train)
score = mlp_reg.score(X_test, y_test)

print(f"MLP Regressor R² Score: {score:.4f}")

plt.figure(figsize=(8, 4))
plt.plot(mlp_reg.loss_curve_)
plt.title('Neural Network Training Loss Curve')
plt.xlabel('Iterations')
plt.ylabel('Loss')
plt.show()
```

**📝 Insight:**
- **Loss Curve**: Should decrease over iterations, indicating the network is learning.
- **R²**: Compare this with the Linear Regression model from Unit II.
